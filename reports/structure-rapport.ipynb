## Structure d'un Rapport de PFE — Niveau Licence

---

### Pages liminaires (avant le contenu)
Ces pages ne sont pas numérotées mais sont essentielles :

**Page de garde** — titre, noms, encadrante, établissement, année universitaire.

**Dédicaces** *(optionnel)* — une page courte, personnelle.

**Remerciements** — remercier l'encadrante, le jury, l'équipe. 10–15 lignes suffisent.

**Résumé** — 150–200 mots en français + 150–200 mots en anglais (Abstract). Décrit le problème, l'approche, et les résultats en quelques lignes.

**Table des matières** — générée automatiquement (Word/LaTeX).

**Liste des figures** et **Liste des tableaux** — idem, automatique.

**Liste des abréviations** — CNN, ML, TL, HAM10000, etc.

---

### Corps du rapport (numéroté)

**Chapitre 1 — Introduction générale**
C'est la vitrine du rapport. Il doit donner envie de lire la suite.
- Contexte général (cancer de la peau, IA en médecine)
- Problématique posée
- Objectifs du projet
- Organisation du rapport (un paragraphe qui annonce les chapitres)

**Chapitre 2 — État de l'art**
C'est la revue de littérature. Tu montres que tu connais le domaine.
- Classification des lésions cutanées (aspect médical)
- Travaux existants sur HAM10000 dans la littérature
- Comparaison des approches CNN utilisées par d'autres chercheurs
- Positionnement de votre travail par rapport à l'existant

**Chapitre 3 — Dataset et Analyse Exploratoire**
- Présentation détaillée de HAM10000
- Analyse des distributions des classes (avec tes figures EDA)
- Analyse des métadonnées tabulaires (âge, sexe, localisation)
- Déséquilibre et biais démographique documentés

**Chapitre 4 — Prétraitement et Préparation des Données**
- Pipeline de preprocessing (redimensionnement, normalisation)
- Gestion du déséquilibre (class_weight, WeightedRandomSampler)
- Data Augmentation (techniques appliquées, justification)
- Encodage des données tabulaires (One-Hot, StandardScaler)

**Chapitre 5 — Modèles et Entraînement**
C'est le cœur technique. Le plus long chapitre.
- CNN from Scratch : architecture, choix des hyperparamètres, résultats
- EfficientNetB0 : Transfer Learning, stratégie Phase A/B, résultats
- Comparaison des deux approches CNN (ton tableau comparatif)
- Random Forest : features, GridSearchCV, Feature Importance

**Chapitre 6 — Évaluation et Résultats**
- Métriques choisies et justification (Recall mélanome en priorité)
- Résultats finaux des trois modèles
- Matrices de confusion, courbes ROC
- Grad-CAM : analyse visuelle des prédictions CNN
- Discussion des résultats (forces, limites, comparaison à la littérature)

**Chapitre 7 — Application Web**
- Architecture globale (Frontend/Backend/IA)
- Présentation de l'interface (captures d'écran)
- API REST : endpoints développés
- Démonstration du flux complet utilisateur

**Chapitre 8 — Conclusion et Perspectives**
- Synthèse du travail réalisé
- Réponse à la problématique posée au chapitre 1
- Limites du projet
- Perspectives d'amélioration (fusion multimodale, données plus diversifiées...)

---

### Pages finales

**Références bibliographiques** — articles, datasets, documentations officielles. Format IEEE ou APA selon votre faculté.

**Annexes** *(si nécessaire)* — code source, résultats supplémentaires, tableaux détaillés.

---

### En résumé visuel

```
Liminaires        Corps du rapport               Fin
─────────   ──────────────────────────────   ─────────
Page garde  Chap.1 Introduction               Références
Dédicaces   Chap.2 État de l'art              Annexes
Remerciements Chap.3 Dataset & EDA
Résumés     Chap.4 Prétraitement
TDM         Chap.5 Modèles ← le plus long
Listes      Chap.6 Évaluation & Résultats
            Chap.7 Application Web
            Chap.8 Conclusion
```

---

### Conseils pratiques pour le rédiger en parallèle

Commence **maintenant** par les chapitres que tu peux déjà écrire sans attendre les résultats finaux — **Chapitre 1, 2, 3 et 4** sont entièrement rédigeables dès aujourd'hui. Le **Chapitre 5** se rédige au fur et à mesure que tu termines chaque modèle. Les **Chapitres 6 et 7** viennent en dernier.
